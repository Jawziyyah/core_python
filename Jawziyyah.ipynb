```python
# SET OPERATIONS.
#7 Union of Sets Union combine two sets
set1={1,2,3}
set2={3,4,5}
result=set1.union(set2)
result
```

```python
#Alternatively use the | (pipe symbol)
result = set1|set2
print(result)
```

```python
# Intersectiion  finds common elements between set.
set1={1,2,3,4}
set2={3,4,5,6}
result= set1.intersection(set2)
print(result)

result=set1&set2
print(result)
```

```python
#Difference () elements in one set  but not in another
# it returns elements that exist in the first set but not in the second.
set1={1,2,3,4}
set2={3,4,5,6}
result=set1.difference(set2)
print(result)

result2=set2.difference(set1)
print(result2)
```

```python
# Symetric_difference (): Elements in either set but not  both
# Returns elements that exist that are in one of the sets but not in both
#It is the opposite of intersection
set1={1,2,3,4}
set2={3,4,5,6}
result=set1.symmetric_difference(set2)
print(result)

#or

result2=set1^set2
print(result2)
```
